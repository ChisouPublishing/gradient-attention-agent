本文で省略した入力を補います。fp16の大きな値を例にします。


```python
import numpy as np
from numpy import exp
x = np.array([11.0, 11.1], dtype=np.float16)
```

```python
# 素朴な実装：fp16 なら x > 11.09 で即 inf → 0/0 = NaN
p = exp(x) / exp(x).sum()

# safe softmax：数学的には同値、数値的には壊れない
m = x.max()                    # パス 1
l = exp(x - m).sum()           # パス 2
p = exp(x - m) / l             # パス 3
```

```python
print("safe softmax:", p)
```
