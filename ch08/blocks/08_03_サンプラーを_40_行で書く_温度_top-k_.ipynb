def generate(model, ids, max_new=256, window=64, **kw):
    eos = model.config.eos_token_id
    eos = torch.tensor(eos if isinstance(eos, list) else [eos])   # EOS が複数のモデルもある
    done = torch.zeros(ids.size(0), dtype=torch.bool)             # 系列ごとの終了フラグ
    past = None
    for _ in range(max_new):
        out = model(ids[:, -1:] if past is not None else ids, past_key_values=past, use_cache=True)
        past = out.past_key_values                  # ★ KV キャッシュ（→ §08）
        nxt = sample(out.logits[:, -1, :], prev_ids=ids[:, -window:], **kw)
        nxt = nxt.masked_fill(done[:, None], int(eos[0]))   # 終わった系列は EOS で埋める
        ids = torch.cat([ids, nxt], dim=-1)
        done |= torch.isin(nxt.squeeze(-1), eos)
        if done.all():                              # ★ バッチ全体が終わったら止める
            break
    return ids
