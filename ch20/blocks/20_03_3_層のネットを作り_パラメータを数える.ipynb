```python
# 付録 A-3：nn.Module でモデルを作る
import torch
from torch import nn

class MLP(nn.Module):
    def __init__(self):
        super().__init__()                                # 部品は __init__ で作る
        self.flatten = nn.Flatten()                       # (B, 1, 28, 28) → (B, 784)
        self.layers = nn.Sequential(
            nn.Linear(28 * 28, 256), nn.ReLU(),
            nn.Linear(256, 128), nn.ReLU(),
            nn.Linear(128, 10),                           # 10 種類のロジット（softmax は損失の側で）
        )
    def forward(self, x):                                 # 計算の順番は forward に書く
        return self.layers(self.flatten(x))

torch.manual_seed(0)
model = MLP()
print(model)
n = sum(p.numel() for p in model.parameters())
print("パラメータ数", f"{n:,}", " 内訳", [tuple(p.shape) for p in model.parameters()])
logits = model(torch.rand(3, 1, 28, 28))
print("出力の形", tuple(logits.shape), " 確率にすると各行の合計", logits.softmax(dim=1).sum(dim=1).tolist())
```
