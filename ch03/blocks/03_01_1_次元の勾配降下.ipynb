```python
# 1 次元の勾配降下：L(w) = (w − 2)² + 1、初期値 w0 = 7
def grad(w):
    return 2 * (w - 2)              # dL/dw

def run(eta, w0=7.0, steps=6):
    w, rows = w0, []
    for t in range(steps + 1):
        rows.append((t, w, abs(w - 2)))
        w = w - eta * grad(w)       # w ← w − η·dL/dw
    return rows

for eta in (0.1, 0.5, 0.8, 1.05):
    print(f"η={eta}  倍率 1−2η = {1 - 2*eta:+.2f}")
    for t, w, err in run(eta):
        print(f"  step {t}: w = {w:8.3f}   |w−2| = {err:7.3f}")
```
